Q9. Write to the file 100 random strings whose length between 10 and 15.

````python
import random
import string

f = open("random.txt", "w")

for i in range(100):
    length = random.randint(10, 15)
    s = ""

    for j in range(length):
        s = s + random.choice(string.ascii_letters)

    print(s)          # Shows output on screen
    f.write(s + "\n") # Writes to file

f.close()
````

Q10. Write to the file all prime numbers between 600 and 800.

````python
f = open("prime.txt", "w")

for num in range(600, 801):
    count = 0

    for i in range(1, num + 1):
        if num % i == 0:
            count = count + 1

    if count == 2:
        print(num)
        f.write(str(num) + "\n")

f.close()
````

Q12. WAP to sort following number of elements in a list, calculate time taken and plot the graph.    
Number of elements in list
5k
Time Taken
T1
10k
15k
T2
T3
20k
25k
T4
T5

````python
import random
import time
import matplotlib.pyplot as plt

elements = [5000, 10000, 15000, 20000, 25000]
time_taken = []

for n in elements:
    L = []

    for i in range(n):
        L.append(random.randint(1, 100000))

    start = time.time()

    L.sort()

    end = time.time()

    t = end - start
    time_taken.append(t)

    print("Number of elements:", n)
    print("Time taken:", t, "seconds")

# Plot graph
plt.plot(elements, time_taken, marker='o')

plt.xlabel("Number of elements in list")
plt.ylabel("Time Taken (seconds)")
plt.title("Sorting Time Analysis")

plt.show()
````

Q13. WAP to create a dictionary of student marks in five subjects and you have to find the student  
having maximum and minimum average marks.

````python
students = {
    "Rahul": [80, 75, 90, 85, 70],
    "Amit": [60, 65, 70, 55, 75],
    "Neha": [90, 95, 85, 92, 88],
    "Priya": [50, 60, 55, 65, 70]
}

max_avg = 0
min_avg = 100
max_student = ""
min_student = ""

for name in students:
    total = 0

    for marks in students[name]:
        total = total + marks

    avg = total / 5

    print(name, "Average Marks:", avg)

    if avg > max_avg:
        max_avg = avg
        max_student = name

    if avg < min_avg:
        min_avg = avg
        min_student = name

print("\nStudent having maximum average marks:")
print(max_student, ":", max_avg)

print("\nStudent having minimum average marks:")
print(min_student, ":", min_avg)
````
